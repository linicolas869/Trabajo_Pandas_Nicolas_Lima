```python
# PUNTO 2
import pandas as pd
import matplotlib.pyplot as plt
import plotly
import seaborn as sns

df = pd.read_csv("datasetnico.csv")

sns.scatterplot(
    data = df,
    x = "Potencia (HP)",
    y = "Emisiones CO2(g/km)"
)

plt.xlabel("Potencia (HP)")
plt.ylabel("Emisiones CO2(g/km)")
plt.title("Potencia vs Emisiones")
plt.show()

print("correlacion:")
cor = df["Potencia (HP)"].corr(df["Emisiones CO2(g/km)"])
print(cor)
```

```python
# PUNTO 8
import pandas as pd
import matplotlib.pyplot as plt
import plotly
import seaborn as sns

df = pd.read_csv("datasetnico.csv")
filtro = (df["Marca"]=="Audi") & (df["Potencia (HP)"]>400)
audis = df[filtro]
print(audis)

modelos = audis["Modelo"].value_counts()
print(modelos)

modelos.plot(kind="bar")
plt.xlabel("Modelo")
plt.ylabel("Cantidad de autos")
plt.title("Modelos")

plt.show()

```

```python
# PUNTO 9
import pandas as pd
import matplotlib.pyplot as plt
import plotly
import seaborn as sns

df = pd.read_csv("datasetnico.csv")
agrupar = df.groupby("Marca")["Potencia (HP)"].mean()
print(agrupar)
print("¿Qué marca tiene la mayor potencia promedio?")
print(agrupar.max())
```

```python
# PUNTO 10
import pandas as pd
import matplotlib.pyplot as plt
import plotly
import seaborn as sns

df = pd.read_csv("datasetnico.csv")
tabla_eficiencia = pd.pivot_table(
    df,
    index = "Marca",
    columns = "Combustible",
    values = "Eficiencia de gas",
    aggfunc = "mean"
)

print(tabla_eficiencia)
```
